# 01 – The four Bell states

```python
# Imports: Statevector gives exact amplitudes (no sampling), Aer samples measurements
from qiskit import QuantumCircuit, transpile
from qiskit.quantum_info import Statevector
from qiskit_aer import AerSimulator
sim = AerSimulator()
```

```python
# Function that builds each Bell state: start from |00>, optionally flip bits (X), then H + CNOT
def bell(which):
    qc = QuantumCircuit(2)
    if which in ("psi+", "psi-"): qc.x(1)   # flip qubit 1 to get the |01> family
    if which in ("phi-", "psi-"): qc.x(0)   # flip qubit 0 so H gives a minus sign
    qc.h(0)      # superposition on qubit 0
    qc.cx(0, 1)  # entangle
    return qc
```

```python
# Print the exact statevector of each Bell state (amplitudes for |00>,|01>,|10>,|11>)
import numpy as np
for name in ["phi+", "phi-", "psi+", "psi-"]:
    sv = Statevector(bell(name))
    print(name, np.round(sv.data, 3))
```

```python
# Measure each state 1000 times: all give perfectly correlated (or anti-correlated) outcomes
for name in ["phi+", "phi-", "psi+", "psi-"]:
    qc = bell(name); qc.measure_all()   # add measurement on all qubits
    print(name, sim.run(transpile(qc, sim), shots=1000).result().get_counts())
```
