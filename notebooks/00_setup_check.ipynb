# 00 – Setup check (ideal Bell state)

```python
# Print library versions to confirm the environment works
import qiskit, qiskit_aer
print("qiskit", qiskit.__version__, "| qiskit-aer", qiskit_aer.__version__)
```

```python
# Build a Bell circuit: H puts qubit 0 in superposition, CNOT entangles qubit 1 with it
from qiskit import QuantumCircuit, transpile
qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])  # measure both qubits into classical bits
print(qc.draw())
```

```python
# Run on the ideal (noise-free) Aer simulator for 1000 shots; expect ~50/50 of '00' and '11'
from qiskit_aer import AerSimulator
sim = AerSimulator()
counts = sim.run(transpile(qc, sim), shots=1000).result().get_counts()
print(counts)
assert set(counts) <= {"00", "11"} and abs(counts.get("00", 0) - 500) < 100
```
